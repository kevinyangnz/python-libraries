```python
import numpy as np
import matplotlib.pyplot  as plt
```

### Basics of NumPy Arrays

```python
# Initialize NumPy array consisting of 3 zeros
np.zeros(3)
```

```python
# Empty NumPy array
np.empty(5)
```

```python
# NumPy array with 5 ones
x = np.ones(5)
print(x)

# Type of array
print("Type of array:", type(x))

# Type of each element
print("Type of elements:", type(x[0]))

# Shape of x
print("Shape of array:", x.shape)
```

```python
# Creating NumPy array with Python list
num_list = [1, 2, 3, 4, 5]
np_list = np.array(num_list)
print(np_list)
```

```python
# From 10 to 20 with 4 elements
range_array = np.linspace(10, 20, 5)
print(range_array)
```

### Math Functions

```python
# Addition
add_array = np.array([1, 2, 3, 4, 5])
print(add_array + 10)

# Subtraction
subtract_array = np.array([10, 20, 30, 40, 50])
print(subtract_array - 5)

# Multiplication
multiply_array = np.array([1, 2, 3, 4, 5])
print(multiply_array * 5)

# Division
divide_array = np.array([3, 6, 9, 12, 15, 18, 21, 24, 27, 30])
print(divide_array / 3)

# Dot product
dot_prod = np.array([4, 8, 12]) @ np.array([3, 6, 9]) 
print(dot_prod)
```

```python
# Math functions with multiple NumPy arrays

# Adding two arrays together
addition = add_array + subtract_array
print(addition)

# Multiplying two arrays together
multiplication = np.array([2, 4, 6]) * np.array([5, 10, 15])
print(multiplication)
```

```python
# Boolean masking and indexing

# True if element divides 3
mask_array = np.array([2, 3, 4, 6, 8, 12])
print(mask_array % 3 == 0)

# Subsets the array to only include True elements
print(mask_array[mask_array % 3 == 0])
```

### Image Slicing

Several image libraries in Python utilise NumPy arrays.

```python
# Read in the image
img = plt.imread('auckland_cbd.jpg')
plt.imshow(img)

# Get the image shape
print("Image shape:", img.shape)
```

```python
# Reduce the pixel length and width by 1/10
plt.imshow(img[::10, ::10]);
```

```python
# Mirrored image
plt.imshow(img[::, ::-1]);
```

```python
# Flip the image upside down
plt.imshow(img[::-1, ::]);
```

```python
# Pixels above 80 set to 255 (white), pixels 80 and below set to 0 (black)
img_mask = np.where(img > 80, 255, 0)
plt.imshow(img_mask);
```

```python
# Zoom in on the Sky Tower
plt.imshow(img[1250:2500, 2500:3500]);
```
